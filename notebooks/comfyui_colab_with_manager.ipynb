### Run ComfyUI with cloudflared (Recommended Way)




```python
!pwd
!ls main.py
```

```python
# Install cloudflared
!wget -q -P ~ https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb
!dpkg -i ~/cloudflared-linux-amd64.deb

import subprocess
import threading
import time
import socket
import urllib.request

PORT = 8188

def wait_for_port(host, port, timeout=120):
    start = time.time()
    while True:
        if time.time() - start > timeout:
            raise TimeoutError(f"Port {port} not open after {timeout}s")
        time.sleep(0.5)
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        result = sock.connect_ex((host, port))
        sock.close()
        if result == 0:
            return True

def cloudflared_tunnel(port):
    # Wait until ComfyUI is up
    wait_for_port("127.0.0.1", port)
    print("\nComfyUI detected on port", port, "- launching cloudflared")
    p = subprocess.Popen(
        ["cloudflared", "tunnel", "--url", f"http://127.0.0.1:{port}"],
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE
    )
    for line in p.stderr:
        l = line.decode()
        if "trycloudflare.com " in l:
            print("This is the URL to access ComfyUI:", l[l.find("http"):], end='')

def test_local():
    wait_for_port("127.0.0.1", PORT)
    try:
        resp = urllib.request.urlopen(f"http://127.0.0.1:{PORT}/", timeout=5)
        print("\nLocal check status:", resp.status)
    except Exception as e:
        print("\nLocal check error:", e)

# Start cloudflared in background
threading.Thread(target=cloudflared_tunnel, daemon=True, args=(PORT,)).start()

# Start ComfyUI
!python /content/ComfyUI/main.py --listen 0.0.0.0 --port {PORT}

# Optional: after server is up, test locally
test_local()
```
